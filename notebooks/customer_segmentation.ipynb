```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/retail_customers.csv')
df.head()
```

```python
features = df[['TotalSpend', 'PurchaseFrequency', 'LoyaltyScore',
               'CategorySpend_Electronics', 'CategorySpend_Clothing',
               'CategorySpend_Grocery']]
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
df['Segment'] = kmeans.fit_predict(scaled_features)
```

```python
df['Segment'].value_counts()
```

```python
df.to_csv('../tableau/customer_segments.csv', index=False)
```
